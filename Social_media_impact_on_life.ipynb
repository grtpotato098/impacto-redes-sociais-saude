```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('Social_media_impact_on_life.csv')
```

```python
df.columns
```

```python

fig,ax = plt.subplots(1,2, figsize = (14,5))
fig.suptitle('Sleep quality X Social media time usage', fontname = 'sans-serif', fontweight='bold', fontsize = 16)
sns.histplot(data = df, x = 'Sleep_Quality_Score', bins = 5, ax = ax[0])
ax[0].set_xlabel('Sleep Quality Score', fontname='sans-serif',  fontsize = 12, color = 'dimgray')
ax[0].set_ylabel('Contagem', fontname='roboto',  fontsize = 12, color = 'dimgray')



ax[1].scatter(x = df['Daily_Usage_Hours'], y =  df['Sleep_Quality_Score'])
ax[1].set_xlabel("Daily time usage", fontname='sans-serif', fontsize = 12, color = 'dimgray')
ax[1].set_ylabel("Sleep quality" , fontname='sans-serif', fontsize = 12, color = 'dimgray')

plt.tight_layout()
plt.show()
```

```python
fig,ax=plt.subplots(1,2, figsize = (14,5))
fig.suptitle('Estresse e Saúde Mental X Uso Diário de Redes Sociais', fontweight = 'bold', fontsize = 16, fontname = 'sans-serif')
ax[0].scatter(x=df['Daily_Usage_Hours'], y=df['Perceived_Stress_Score'])
ax[0].set_xlabel('Uso Diário em Horas', fontname = 'sans-serif', color = 'Dimgray')
ax[0].set_ylabel('Pontuação de Estresse', fontname = 'sans-serif', color = 'Dimgray')


ax[1].scatter(x=df['Daily_Usage_Hours'], y=df['Mental_Health_Index'])
ax[1].set_xlabel('Uso Diário em Horas', fontname = 'sans-serif', color = 'Dimgray')
ax[1].set_ylabel('Indice de Saúde Mental', fontname = 'sans-serif', color = 'Dimgray')


plt.tight_layout()
plt.savefig('scatter_duplo.png')
plt.show()
```

```python
plt.figure(figsize = (10,8))
sns.boxplot(
    data=df, 
    x='Daily_Usage_Hours', 
    y=df['Sleep_Quality_Score'].astype('category')
)

plt.title('Qualidade de Sono X Uso Diário de Redes Sociais', fontname = 'sans-serif', fontsize = 15, fontweight = 'bold' )
plt.xlabel('Uso Diário em Horas', fontname = 'sans-serif', fontsize = 12, color = 'dimgray')
plt.ylabel('Pontuação Qualidade de Sono', fontname = 'sans-serif', fontsize = 12, color = 'dimgray')

sns.despine()

plt.tight_layout()
plt.savefig('boxplot_horizontal.png')
plt.show()
```
